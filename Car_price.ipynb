```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
df.columns
```

```python
df.rename(columns={"msrp" : "Price"} , inplace=True)
```

```python
df.columns
```

```python
df.describe()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df.Price, bins=100)
plt.show()
```

```python
df['Price'] = np.log1p(df.Price)
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df.Price, bins=100)
plt.show()
```

```python
print(df.isnull().sum())
```

```python
df = df.drop(["market_category"], axis=1)
```

```python
df ['number_of_doors'] = df ['number_of_doors'].fillna(df['number_of_doors'].mean())
df['engine_hp'] = df['engine_hp'].fillna(df['engine_hp'].mean())
df['engine_cylinders'] = df['engine_cylinders'].fillna(df['engine_cylinders'].mean())
```

```python
df = df.drop(['engine_fuel_type'],axis=1)
```

```python
df.isnull().sum()
```

```python
df = df.drop(['transmission_type', 'driven_wheels', 'vehicle_style'], axis=1)
```

```python
df.columns
```

```python
df.head()
```

```python
df = pd.get_dummies(df, columns=["make", 'model', 'vehicle_size'])
```

```python
df.head()
```

```python
x = df.drop(['Price'], axis=1)
y = df['Price']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train ,x_test , y_train ,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)
```

```python
from sklearn.metrics import mean_squared_error , r2_score
```

```python
y_test_pred = model.predict(x_test)
mse = mean_squared_error(y_test, y_test_pred)
rmse = np.sqrt(mse)
print(rmse)
```

```python
y_train_pred = model.predict(x_train)
r2 = r2_score(y_train, y_train_pred)
print(r2)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
tree_model = DecisionTreeRegressor()
tree_model.fit(x_train, y_train)
```

```python
y_tree_pred = tree_model.predict(x_train)
mse = mean_squared_error(y_train, y_tree_pred)
rmse = np.sqrt(mse)
print(rmse)
```

```python
r2 = r2_score(y_train, y_tree_pred)
print(r2)
```

```python
y_test_pred = tree_model.predict(x_test)
mse = mean_squared_error(y_test , y_test_pred)
rmse = np.sqrt(mse)
print(rmse)
```

```python
r2_test = r2_score(y_test, y_test_pred)
print(r2_test)
```
